# Gatekeeper AI — OpenAI Integration ExampleThis notebook demonstrates how to integrate Gatekeeper with OpenAI's Python SDK.

```python
from gatekeeper import Gatekeepergk = Gatekeeper(    api_key="sk-your-openai-key",    knowledge_base=[        "Our refund policy: 30-day returns, original packaging required.",        "Shipping: UPS Ground (3-5 days), FedEx Express (1-2 days).",        "Product supports Python 3.10+, Node.js 18+.",    ],    brand_guidelines="Professional tone. No slang. Use 'we' not 'I'.",    policy_rules={        "financial": [r"guaranteed returns", r"risk.?free"],    },    mode="flag",)
```

```python
# Example 1: Normal query (should pass)response, scan = gk.chat.completions.create(    model="gpt-4o-mini",    messages=[{"role": "user", "content": "What's the shipping time for UPS Ground?"}])print(f"Risk: {scan.risk_score} | Severity: {scan.severity.value}")print(f"Response: {response.choices[0].message.content[:100]}")
```

```python
# Example 2: Risky query (may trigger scoring)response, scan = gk.chat.completions.create(    model="gpt-4o-mini",    messages=[{"role": "user", "content": "Tell me about your refund policy and investment returns"}])print(f"Risk: {scan.risk_score} | Severity: {scan.severity.value}")for name, dim in scan.dimensions.items():    print(f"  {name}: {dim.score} - {dim.details}")
```
